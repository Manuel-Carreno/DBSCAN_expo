## Aprendizaje no Supervisado y Clustering

### ¿Que es el aprendizaje no supervisado?
Es un algoritmo utilizado para analizar y agrupar conjuntos de datos no etiquetados. Estos algoritmos, desccubren patrones o agrupaciones en los datos sin necesidad de intervención humana.

**Casos de Uso:**
- Computer Vision
- Motores de Recomendación
- Secciones de Noticias
- Detección de Anomalias

### ¿Que es Clustering?
Es una tecnica que agrupa datos no etiquetados en función de las similitudes o diferencias. Los algoritmos de clustering se utilizan para procesar objetos de datos sin procesar y sin clasificar en grupos representados por estructuras o patrones en la información

----

## ¿Que es DBSCAN?
Es un algoritmo de aprendizaje no supervisado que agrupa puntos de datos basandose en la densidad de estos datos, no en centroides (a comparación de K-Means). Es ideal para identificar formas complejas y detectar valores atipicios (outliers o ruido) automaticamente.

A diferencia de K-Means:
- ✅ **K-Means usa centroides mientras que DBSCAN usa densidad para agrupar puntos**
- ✅ **DBSCAN puede identificar grupos de formas irregulares, mientras que K-Means solo logra identificar los grupos que son esfericos**
- ✅ **DBSCAN no necesita que se especifique el número de clústeres antes de iniciar, mientras que K-Means sí lo requiere**

----

## Conceptos Fundamentales

DBSCAN tiene varios parámetros, pero los dos más importantes son:

#### `eps` (ε) → Distancia máxima entre dos puntos para que se consideren vecinos o cercanos

- Para estimar **`eps`**, se utiliza el k-distance plot:
  - Para cada punto, se calcula la distancia a su **minPts-ésimo vecino**
  - Se ordenan las distancias
  - Se grafica la secuencia de distancias.
  - Se identifica el codo de la curva, es decir, el punto donde haya un cambio brusco


#### `min_samples` (MinPts) → Número mínimo de puntos necesarios para formar una región densa


- No existe una formula exacta para calcular el numero minimo de puntos necesarios para formar una región densa.Sin embargo, comúnmente se utiliza el criterio:

  **`min_samples ≥ dimensión del dataset + 1`**

- Esto se debe a que, para considerar que un punto pertenece a una **región densa**, debe tener suficientes vecinos cercanos dentro de su vecindario.

----

## Tipos de Puntos

![Tipos de puntos en DBSCAN](\img\tipos_puntos.png)

| Tipo de punto | Condición | Descripción / Rol |
|---|---|---|
| 🔵**Punto Núcleo** | El punto P tiene al menos **`minPts`** dentro de un radio **`eps`**| Es la base para formar y expandir clusters. |
| 🟡**Punto Borde** | P tiene menos de **`minPts`** dentro de un radio **`eps`**, pero se encuentra en la vecindad de un punto nucleo| Pertenece a un cluster, pero no lo expande. |
| 🔴**Punto Ruido** | P es un punto de ruido si no es un punto nucleo o punto borde y ademas se encuentra muy lejano de cualquier vecindad | No pertenece a ningun cluster y usualmente se etiqueta como `-1`|



## 📊 Descripción del dataset
El dataset a continuación sera usado para demostrar como funciona DBSCAN. Este dataset, consta de 200 filas y 5 columnas, contiene información demográfica y de comportamiento de compra de clientes en un centro comercial. Para este caso haremos uso de las columnas de *Annual Income* y *Spending Score* que nos permitiran visualizar los grupos e identificar patrones de comportamiento entre clientes

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.neighbors import NearestNeighbors
import numpy as np
```

```python
df = pd.read_csv("data\Mall_Customers.csv")
df.head()
```

### Selección de variables

```python
X = df[["Annual Income (k$)", "Spending Score (1-100)"]].values
```

### Visualización de datos 

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['Annual Income (k$)'], df['Spending Score (1-100)'], s=50, alpha=0.7)
plt.title('Annual Income vs Spending Score')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.show()
```

## 📏Elección de parametros
----

### Elegir `min_samples` y estimar `eps` **con k-distance plot** 

```python
# Definimos el valor de min_samples
min_samples = 3

# Calculamos las distancias al 3er vecino más cercano
nbrs = NearestNeighbors(n_neighbors=min_samples)
nbrs.fit(X)

distances, indices = nbrs.kneighbors(X)

# Tomamos la distancia al vecino número k (última columna)
k_distances = np.sort(distances[:, -1])

# Graficamos el k-distance plot
plt.figure(figsize=(8, 5))
plt.plot(k_distances)
plt.title("k-distance plot para estimar eps")
plt.xlabel("Puntos ordenados")
plt.ylabel(f"Distancia al {min_samples}° vecino")
plt.grid(alpha=0.3)
plt.show()
```
